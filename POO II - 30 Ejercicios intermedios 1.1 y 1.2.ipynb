### Ejercicio 1 · Factura con descuento e IVA ★★

Pida el nombre de un producto, la cantidad y el precio unitario. Calcule:
- Subtotal = cantidad × precio
- Descuento del 10 % sobre el subtotal
- IVA del 15 % sobre (subtotal − descuento)
- Total

Muestre la factura con los valores **alineados a la derecha** en 10 espacios y 2 decimales:
```
Producto: Teclado
Subtotal:      90.00
Descuento:      9.00
IVA:           12.15
TOTAL:         93.15
```

````python
producto = input("Nombre del producto: ")
cantidad = int(input("Cantidad: "))
precio = float(input("Precio unitario: "))

subtotal = cantidad * precio
descuento = subtotal * 0.10
iva = (subtotal - descuento) * 0.15
total = subtotal - descuento + iva

print(f"Producto: {producto}")
print(f"Subtotal:  {subtotal:>10.2f}")
print(f"Descuento: {descuento:>10.2f}")
print(f"IVA:       {iva:>10.2f}")
print(f"TOTAL:     {total:>10.2f}")
````

Hágalo de dos formas: con `** 0.5` y con `math.sqrt`. Muestre el resultado con 3 decimales.

````python
import math

x1 = float(input("x1: "))
y1 = float(input("y1: "))
x2 = float(input("x2: "))
y2 = float(input("y2: "))

# Forma 1
distancia_1 = ((x2 - x1)**2 + (y2 - y1)**2) ** 0.5
# Forma 2
distancia_2 = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)

print(f"Distancia (potencia): {distancia_1:.3f}")
print(f"Distancia (math): {distancia_2:.3f}")
````

Pida el capital inicial, la tasa anual (%) y los años. Calcule el monto final con interés compuesto, $M = C(1 + t)^n$, y la ganancia. Compárelo con el interés simple ($C \cdot t \cdot n$) y muestre la diferencia, todo con `:,.2f`.

````python
capital = float(input("Capital inicial: "))
tasa_anual = float(input("Tasa anual (%): "))
anios = int(input("Años: "))

tasa_decimal = tasa_anual / 100
monto_compuesto = capital * (1 + tasa_decimal) ** anios
ganancia_compuesto = monto_compuesto - capital

interes_simple = capital * tasa_decimal * anios
diferencia = ganancia_compuesto - interes_simple

print(f"Monto final (compuesto): {monto_compuesto:,.2f}")
print(f"Ganancia (compuesto): {ganancia_compuesto:,.2f}")
print(f"Diferencia vs interés simple: {diferencia:,.2f}")
````

### Ejercicio 22 · Inventario con diccionario ★★

Con `stock = {"laptop": 12, "mouse": 45, "teclado": 30, "monitor": 8}`:
1. Se vendieron 5 mouses: actualice el stock.
2. Llegó un producto nuevo: agregue `"webcam": 15` con `update`.
3. Se dejó de vender el monitor: elimínelo con `pop` y muestre cuántas unidades tenía.
4. Consulte `"impresora"` con `get` para que muestre `0` si no existe.

```python
python_t = {"Ana", "Luis", "María", "Pedro", "Sofía"}
excel_t  = {"Luis", "Carla", "Pedro", "Diego"}
ia_t     = {"Ana", "Pedro", "Diego"}
```
Muestre (ordenado con `sorted`):
1. Quiénes están en Python **y** en Excel.
2. Quiénes están en los tres talleres.
3. Quiénes están en Python pero **no** en IA.
4. Quiénes están en Python **o** en Excel, pero no en ambos (`^`).
5. Si todos los de IA están también en Python (`issubset`).
6. El total de estudiantes distintos.

````python
python_t = {"Ana", "Luis", "María", "Pedro", "Sofía"}
excel_t  = {"Luis", "Carla", "Pedro", "Diego"}
ia_t     = {"Ana", "Pedro", "Diego"}

python_y_excel = sorted(python_t & excel_t)
tres_talleres = sorted(python_t & excel_t & ia_t)
python_no_ia = sorted(python_t - ia_t)
python_o_excel_exclusivo = sorted(python_t ^ excel_t)
ia_es_subconjunto = ia_t.issubset(python_t)
total_distintos = len(python_t | excel_t | ia_t)

print(f"Python y Excel: {python_y_excel}")
print(f"Los 3 talleres: {tres_talleres}")
print(f"Python pero NO IA: {python_no_ia}")
print(f"Python o Excel (no ambos): {python_o_excel_exclusivo}")
print(f"¿Todos los de IA en Python?: {ia_es_subconjunto}")
print(f"Total estudiantes: {total_distintos}")
````

### Ejercicio 29 · Copia o referencia ★★

Ejecute y explique en comentarios cada resultado:
```python
original = {"nombre": "Ana", "saldo": 100}
alias = original
copia = original.copy()
alias["saldo"] = 500
copia["saldo"] = 999
```
1. Muestre `original`, `alias` y `copia`.
2. Compruebe con `is` cuáles son el mismo objeto.
3. Compare con `id()` las direcciones de memoria.

````python
original = {"nombre": "Ana", "saldo": 100}
alias = original
copia = original.copy()

alias["saldo"] = 500
copia["saldo"] = 999

print(f"Original: {original}") # Cambia porque 'alias' modificó la misma referencia.
print(f"Alias: {alias}")
print(f"Copia: {copia}")

mismo_objeto_alias = original is alias
mismo_objeto_copia = original is copia

print(f"Original is Alias: {mismo_objeto_alias}")
print(f"Original is Copia: {mismo_objeto_copia}")
print(f"IDs: Original({id(original)}), Alias({id(alias)}), Copia({id(copia)})")
````

### Ejercicio 30 · Ficha de un socio a partir de texto ★★

Un archivo antiguo guarda cada socio en una sola línea:
`linea = "ana pérez|0104567896|ahorros=1520.75|plazo fijo=5000"`
1. Separe la línea por `"|"`.
2. Cree un diccionario con `nombre` (formato título), `cedula` y `cuentas`, que a su vez es un diccionario `{tipo: saldo}` (separe cada cuenta por `"="` y convierta el saldo a `float`).
3. Muestre la ficha con este formato:
```
Socio: Ana Pérez (0104567896)
Tipos de cuenta: ahorros, plazo fijo
Saldo total: $6,520.75
```

````python
linea = "ana pérez|0104567896|ahorros=1520.75|plazo fijo=5000"

partes = linea.split("|")
cuenta_1 = partes[2].split("=")
cuenta_2 = partes[3].split("=")

socio = {
    "nombre": partes[0].title(),
    "cedula": partes[1],
    "cuentas": {
        cuenta_1[0]: float(cuenta_1[1]),
        cuenta_2[0]: float(cuenta_2[1])
    }
}

lista_tipos = list(socio["cuentas"].keys())
saldo_total = sum(socio["cuentas"].values())

print(f"Socio: {socio['nombre']} ({socio['cedula']})")
print(f"Tipos de cuenta: {lista_tipos[0]}, {lista_tipos[1]}")
print(f"Saldo total: ${saldo_total:,.2f}")
````
