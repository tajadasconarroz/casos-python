## 2. Decidir antes de repetir: `if`, `elif`, `else` (15 min)

Casi todo ciclo necesita una **condición**, así que primero veremos cómo se escribe una decisión en Python.

| Java | Python |
|---|---|
| `if (saldo >= 100) { ... }` | `if saldo >= 100:` |
| `else if (...)` | `elif ...:` |
| Llaves `{ }` marcan el bloque | La **sangría (4 espacios)** marca el bloque |
| `&&`, `\|\|`, `!` | `and`, `or`, `not` |

```python
# Validador de saldo y categoría:
# Si el saldo es mayor o igual a 1000 -> categoría Oro,
# si es mayor o igual a 100 -> categoría Plata,
# de lo contrario -> categoría Básica

saldo = float(input("Ingrese su saldo: "))

if saldo >= 1000:
    categoria = "Oro"
elif saldo >= 100:
    categoria = "Plata"
else:
    categoria = "Básica"

print(f"Su categoría es: {categoria}")

```

### Condiciones compuestas
Python permite **encadenar comparaciones** como en matemáticas: `18 <= edad < 65`. En Java habría que escribir `edad >= 18 && edad < 65`.

```python
# Validador de edad, activo y deuda para solicitar un crédito:
# Si la edad está entre 18 y 64 años, y está activo y no tiene deuda,
# puede solicitar un crédito caso contrario no cumple los requisitos

edad = int(input("Ingrese su edad: "))
activo = True   # Puedes cambiar esto a False para probar
tiene_deuda = False # Puedes cambiar esto a True para probar

if 18 <= edad <= 64 and activo and not tiene_deuda:
    print("Puede solicitar un crédito.")
else:
    print("No cumple los requisitos para solicitar un crédito.")
```

### Expresión condicional (operador ternario)
En Java: `String estado = saldo > 0 ? "con fondos" : "sin fondos";`

```python
# Validador de saldo. si el saldo es mayor a 0 -> con fondos, caso contrario -> sin fondos
saldo = float(input("Ingrese su saldo: "))
estado = "con fondos" if saldo > 0 else "sin fondos"
print(estado)
```

> ⚠️ **Error típico de quien viene de Java:** olvidar los dos puntos `:` al final del `if`, o mezclar espacios y tabulaciones en la sangría. Python muestra `SyntaxError` o `IndentationError`.

## 3. El ciclo `while` (20 min)

`while` repite un bloque **mientras** la condición sea verdadera. Se usa cuando **no sabemos cuántas veces** vamos a repetir.

| Java | Python |
|---|---|
| `while (i <= 5) { ...; i++; }` | `while i <= 5:` … `i += 1` |
| `i++` | No existe: se escribe `i += 1` |
| `do { ... } while (...)` | No existe: se usa `while True` con `break` |

```python
# Contador: cuenta de 1 a 5
for i in range(1, 6):
    print(f"Vuelta {i}")
```

### Acumulador
Un **acumulador** guarda una suma (o un producto) que crece en cada vuelta. Se inicia en `0` para sumar y en `1` para multiplicar.

**Ejemplo:** un socio deposita $50 cada mes. ¿En cuántos meses llega a $400?

```python
# Acumulador: suma de 50 hasta llegar a 400 e imprime el saldo y la cantidad de meses que se tardó en llegar a ese saldo
saldo = 0
meses = 0

while saldo < 400:
    saldo += 50
    meses += 1

print(f"Llega a ${saldo:.2f} en {meses} meses")
```

### Validar datos con `while`
Un uso muy común: **pedir un dato hasta que sea válido**. Ejecute la celda y pruebe con un monto negativo.

```python
# Validador de monto a depositar:
# solicita al usuario un monto a depositar y valida que sea mayor que cero.
# Si el monto es menor o igual a cero, solicita nuevamente el monto hasta que sea válido.
# Finalmente, imprime el monto registrado con dos decimales.

monto = 0
while monto <= 0:
    monto = float(input("Ingrese el monto a depositar: "))
    if monto <= 0:
        print("El monto debe ser mayor que cero. Intente de nuevo.")

print(f"Monto registrado: ${monto:.2f}")
```

### `while True` + `break`: el «do-while» de Python
Cuando el ciclo debe ejecutarse **al menos una vez**, se escribe `while True` y se sale con `break`. Escriba `salir` para terminar.

```python
# Solicitar al usuario ingrese una nota
# si la nota es mayor o igual cero y menor o igual a 10 nuevamente solicitamos
# Si el usuario ingresa una nota que es igual a 10 termino el ciclo repetitivo
# Si no solo imprime la nota
# Imprimir la nota

while True:
    nota = float(input("Ingrese una nota (entre 0 y 10): "))
    
    if nota == 10:
        print(nota)
        break
    elif 0 <= nota <= 10:
        print(nota)
    else:
         print("Nota fuera de rango. Intente nuevamente.")

```

> ⚠️ **Ciclo infinito:** si la condición nunca se vuelve falsa, el ciclo no termina. En Jupyter se detiene con el botón **■ (Interrumpir)** o con `I`, `I`.

## 4. El ciclo `for` con `range` (25 min)

`for` se usa cuando **sabemos cuántas veces** repetir o queremos recorrer una colección. `range` genera la secuencia de números.

| Java | Python | Valores |
|---|---|---|
| `for (int i = 0; i < 5; i++)` | `for i in range(5):` | 0, 1, 2, 3, 4 |
| `for (int i = 1; i <= 12; i++)` | `for i in range(1, 13):` | 1 … 12 |
| `for (int i = 0; i < 10; i += 2)` | `for i in range(0, 10, 2):` | 0, 2, 4, 6, 8 |
| `for (int i = 5; i > 0; i--)` | `for i in range(5, 0, -1):` | 5, 4, 3, 2, 1 |

> 📌 Igual que en el slicing, **el final no se incluye**: `range(1, 13)` llega hasta 12.

```python
# Imprimir los numeros del 0 al 4 en una sola línea, separados por espacios, y luego imprimir "Fin de ejecución" en una nueva línea.
for i in range(5):
    print(i, end=" ")
print("\nFin de ejecución")
```

```python
# Imprimir los números del 1 al 12 en una sola línea, separados por espacios, y luego imprimir "Fin de ejecución" en una nueva línea.
for i in range(1, 13):
    print(i, end=" ")
print("\nFin de ejecución")
```

```python
# Imprimir los números del 0 al 10 (excluyendo el 10) en saltos de 2, separados por espacios, y luego imprimir "Fin de ejecución" en una nueva línea.
for i in range(0, 10, 2):
    print(i, end=" ")
print("\nFin de ejecución")
```

```python
# Imprimir los números del 5 al 1 en una sola línea,
# separados por espacios, y luego imprimir "¡Despegue!" en una nueva línea.
for i in range(5, 0, -1):
    print(i, end=" ")
print("\n¡Despegue!")
```
